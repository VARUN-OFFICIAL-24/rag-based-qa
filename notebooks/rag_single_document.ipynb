```python
from langchain.document_loaders import PyPDFLoader

def load_documents():
    file_path = r"C:\Users\example.pdf"
    document_loader = PyPDFLoader(file_path)  # Use PyPDFLoader for a single file
    return document_loader.load()
```

```python
documents = load_documents()
print(documents[0])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain.schema.document import Document
```

```python
def split_documents(documents: list[Document]):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=800,
        chunk_overlap=80,
        length_function=len,
        is_separator_regex=False,
    )
    return text_splitter.split_documents(documents)
```

```python
documents=load_documents()
chunks = split_documents(documents)
```

```python
from langchain.text_splitter import CharacterTextSplitter
from langchain.vectorstores import FAISS
```

```python

from PyPDF2 import PdfReader
file_path = r"C:\Users\Downloads\example.pdf"
doc_reader = PdfReader(file_path)

```

```python
doc_reader
```

```python

raw_text=""

for i, page in enumerate(doc_reader.pages):
    text = page.extract_text()
    if text:
        raw_text+=text

len(raw_text)
```

```python
raw_text[:1000]
```

```python
text_splitter = CharacterTextSplitter(
    separator = "\n",
    chunk_size=1000,
    chunk_overlap = 200,
    length_function = len

)

texts = text_splitter.split_text(raw_text)
```

```python
texts[20]
```

```python
len(texts)
```

```python
embeddings
```

```python
docsearch = FAISS.from_texts(texts, embeddings)
```

```python
docsearch.embedding_function
```

```python
query = "how does GPT-4 change social media?"
docs = docsearch.similarity_search(query)
```

```python
from langchain.chains import RetrievalQA

# set up FAISS as a generic retriever
retriever = docsearch.as_retriever(search_type="similarity", search_kwargs={"k":4})

# create the chain to answer questions
rqa = RetrievalQA.from_chain_type(llm=OllamaLLM(model="llama3.2"),
                                  chain_type="stuff",
                                  retriever=retriever,
                                  return_source_documents=True)
```

```python
rqa("What is OpenAI?")
```
